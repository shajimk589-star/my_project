````python
import gradio as gr
import torch
import torchvision.transforms as transforms
from PIL import Image
import os
import sys
import cv2
import numpy as np

if "." not in sys.path:
    sys.path.append(".")

from model import DeepfakeDetectorResNet

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = DeepfakeDetectorResNet(pretrained=False)
modelpath = "model_resnet.pth" # Corrected filename

if not os.path.exists(modelpath) and os.path.exists(modelpath + ".zip"):
    import zipfile
    with zipfile.ZipFile(modelpath + ".zip", "r") as zip_ref:
        zip_ref.extractall()
    if not os.path.exists(modelpath):
        if os.path.exists("modelresnet/modelresnet.pth"):
            modelpath = "modelresnet/modelresnet.pth"
        elif os.path.exists("modelresnet.pth"):
            modelpath = "modelresnet.pth"
        else:
            raise FileNotFoundError(f"Model file not found at {modelpath} or extracted paths.")

model.load_state_dict(torch.load(modelpath, map_location=device))
model.to(device)
model.eval()

preprocess = transforms.Compose([
    transforms.Resize(256),
    transforms.CenterCrop(224),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225]),
])

def predict_single_frame(image: Image.Image) -> float:
    if image.mode != "RGB":
        image = image.convert("RGB")
    img_tensor = preprocess(image).unsqueeze(0).to(device)
    with torch.no_grad():
        output = model(img_tensor)
        prediction = torch.sigmoid(output).item()
    return prediction

def predict_deepfake_video(video_path: str) -> str:
    if video_path is None:
        return "Please upload a video."

    cap = cv2.VideoCapture(video_path)
    if not cap.isOpened():
        return "Error: Could not open video file."

    frame_predictions = []
    frame_count = 0
    frame_skip = 10

    while True:
        ret, frame = cap.read()
        if not ret:
            break

        frame_count += 1
        if frame_count % frame_skip != 0:
            continue

        pil_image = Image.fromarray(cv2.cvtColor(frame, cv2.COLOR_BGR2RGB))
        prediction = predict_single_frame(pil_image)
        frame_predictions.append(prediction)

    cap.release()

    if not frame_predictions:
        return "No frames processed. Video might be too short or corrupted."

    avg_prediction = np.mean(frame_predictions)
    deepfake_frames = sum(1 for p in frame_predictions if p > 0.5)
    real_frames = len(frame_predictions) - deepfake_frames

    result = f"""Processed {len(frame_predictions)} frames.
Average deepfake probability: {avg_prediction:.4f}
"""

    if avg_prediction > 0.5:
        result += f"Overall result: DEEPFAKE ({deepfake_frames} frames flagged)."
    else:
        result += f"Overall result: REAL ({real_frames} frames flagged)."

    return result

with gr.Blocks(theme=gr.themes.Soft(), css="""
    .title {text-align: center; font-size: 34px; font-weight: 700; margin-bottom: 0;}
    .subtitle {text-align: center; font-size: 16px; color: #666; margin-top: 0;}
    .card {padding: 18px; border-radius: 18px; box-shadow: 0 4px 20px rgba(0,0,0,0.06);}
""") as demo:
    gr.Markdown("<h1 class='title'>Video Deepfake Detection</h1>")
    gr.Markdown("<p class='subtitle'>Upload a video and analyze it frame by frame for deepfake detection.</p>")

    with gr.Row():
        with gr.Column(scale=1):
            gr.Markdown("### Upload Video")
            video_input = gr.Video(label="Choose a video", sources=["upload"], format="mp4")
            run_btn = gr.Button("Detect Deepfake", variant="primary")
        with gr.Column(scale=1):
            gr.Markdown("### Result")
            output_text = gr.Textbox(label="Analysis output", lines=8, interactive=False)

    gr.Markdown("### How it works")
    gr.Markdown(
        "- The app samples every 10th frame.\n"
        "- Each frame is passed through the trained model.\n"
        "- The final result is based on the average prediction."
    )

    run_btn.click(fn=predict_deepfake_video, inputs=video_input, outputs=output_text)

demo.launch(share=True, debug=True)
````

# AI Deepfake Video Detection

An AI-powered deepfake video detection system designed to identify manipulated and AI-generated videos using a hybrid spatial-temporal deep learning approach.

## 📌 Overview

The rapid development of Generative AI, GANs, diffusion models, and neural rendering has made it increasingly difficult to distinguish real videos from manipulated content.

This project presents an automated Deepfake Video Detection pipeline that analyzes both:

- Spatial visual artifacts
- Frequency-domain abnormalities
- Temporal inconsistencies across video frames

The system classifies videos as **Real** or **Fake** and is designed with real-world video compression and deployment scenarios in mind.

## 🎯 Objectives

The main objectives of this project are:

- Detect AI-generated and manipulated videos automatically.
- Extract facial regions from video frames.
- Identify spatial artifacts using deep learning models.
- Analyze high-frequency manipulation patterns using FFT/DCT.
- Detect temporal inconsistencies using LSTM and 3D-CNN models.
- Improve robustness against H.264/H.265 video compression.
- Evaluate the system using Accuracy, AUC-ROC, F1-Score and EER.
- Explore real-time inference and production deployment.
